**Before you start**, you need a Google Maps API key. It is free and takes about
ten minutes, once:

1. Create a project: https://console.cloud.google.com/projectcreate — call it anything.
2. Turn on billing: https://console.cloud.google.com/billing → *Link a billing account*.

4. Make the key: https://console.cloud.google.com/apis/credentials →
   *Create credentials* → *API key*. Copy it.

Then run the three cells below in order — click the ▶ button on each one.

### If something goes wrong

**`403` or `PERMISSION_DENIED`** — the Places API (New) is not switched on for the
project the key belongs to, or the key is restricted to a different API. Go back to
step 3 of the setup above.

**`REQUEST_DENIED` / billing** — billing is not linked to that project yet.

**Nothing downloads at the end** — some browsers block multiple downloads. Look in
the file panel on the left (the folder icon), right-click each file, *Download*.

### A note on what is in here

Company name, category, phone, website, address, county, rating, review count and
a map link. **There is no email column** — the Google Places API has no email
field at all, so no tool built on it can produce one. The website column is the
way in: the contact form or published `info@` address is one click from there.
